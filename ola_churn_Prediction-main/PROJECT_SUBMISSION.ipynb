```python
def calculate_roi_of_retention():
    """Calculate return on investment of churn prediction model"""
    
    # Costs
    model_development = 50_000  # Engineering, data science
    infrastructure = 10_000     # Deployment, monitoring
    intervention_cost = 500     # Per driver (call, bonus, etc.)
    
    # Benefits
    num_drivers_at_risk = 200   # Identified by model
    intervention_rate = 0.76    # Success rate
    prevented_churns = num_drivers_at_risk * intervention_rate
    
    # Economic impact per retained driver
    customer_lifetime_value = 5_000  # Revenue from driver
    acquisition_cost = 2_000         # Cost to recruit replacement
    
    prevented_churn_value = prevented_churns * (customer_lifetime_value + acquisition_cost)
    total_intervention_cost = num_drivers_at_risk * intervention_cost
    
    net_benefit = prevented_churn_value - (model_development + infrastructure + total_intervention_cost)
    roi = (net_benefit / (model_development + infrastructure)) * 100
    
    return {
        "net_benefit": net_benefit,
        "roi_percent": roi,
        "payback_period_months": (model_development + infrastructure) / (net_benefit / 12)
    }
```

### Challenge 1: Feature Shape Mismatch Error

**Problem Statement:**
```
ValueError: Feature shape mismatch, expected: 43, got 44
ValueError: columns are missing: {'Unnamed: 0'}
```

Occurred 20+ times during dashboard testing.

**Root Cause Analysis:**
1. Raw CSV had `Unnamed: 0` (pandas index column) with 15 features
2. Feature engineering script dropped `Unnamed: 0` correctly
3. Preprocessing script did NOT drop `Unnamed: 0` → preprocessor fitted on 15 features
4. Model trained on 43 features (12 numeric + 31 City one-hot)
5. Old pickled preprocessor (Dec 10) still had 15 input features
6. `main.py` was NOT saving preprocessor.pkl file → old version kept being loaded

**Attempted Solutions:**
- ❌ Add `'Unnamed: 0'` to dashboard manually → Breaks when dropped by preprocessor
- ❌ Filter columns in dashboard → Temporary fix, not root cause
- ❌ Modify only dashboard.py → Doesn't work since preprocessor was old

**Final Solution:**
```python
# 1. Fixed preprocessing_06.py (Line 25)
drop_cols = ["Unnamed: 0", "Driver_ID", "Dateofjoining", "LastWorkingDate", "MMM-YY"]
df.drop(columns=[c for c in drop_cols if c in df.columns], inplace=True)

# 2. Added to main.py (After line 66)
preprocessor_path = "models/preprocessor.pkl"
with open(preprocessor_path, "wb") as f:
    pickle.dump(preprocessor, f)

# 3. Regenerated all artifacts
python main.py  # Preprocessor now has 14 input features
```

**Learning:**
- Pickled artifacts must be regenerated when training code changes
- File modification timestamps are critical debugging tools
- Missing save operations can cascade through entire pipeline

### Challenge 2: Preprocessor Structure Complexity

**Problem:**
```python
AttributeError: 'Pipeline' object has no attribute 'categories_'
```

The categorical transformer was a nested Pipeline, not a direct OneHotEncoder.

**Root Cause:**
```python
# In preprocessing_06.py, cat_features use a Pipeline:
cat_pipe = Pipeline([("ohe", OneHotEncoder(...))])

ohe_cats = preprocessor.named_transformers_["cat"].categories_  # ❌ Wrong!
```

**Solution:**
```python

cat_pipeline = preprocessor.named_transformers_["cat"]
ohe = cat_pipeline.named_steps["ohe"]  # Navigate through Pipeline
ohe_cats = ohe.categories_  # ✅ Now works!

city_cats = list(ohe_cats[0])  # City categories
ym_cats = list(ohe_cats[1])    # year_month categories
```

**Learning:**
- Always account for nested estimator structures in Pipelines
- Test artifact loading separately before integration
- Document expected structure explicitly

### Challenge 3: SHAP Computation Overhead

**Problem:**
- KernelExplainer with 200 samples took 30+ minutes
- Dashboard became unresponsive
- User experience degraded significantly

**Root Cause:**
- KernelExplainer creates many model evaluations (O(m*n) complexity)
- 200 samples × 43 features × multiple iterations = millions of evaluations
- LightGBM warnings on each evaluation slowed computation

**Optimization Process:**
```python
# Before: Very Slow
explainer = shap.KernelExplainer(model.predict_proba, X_transformed)
shap_vals = explainer.shap_values(X_transformed)  # 200 samples → 30 mins

# After: Optimized
background = shap.sample(X_transformed, 20)  # Smaller background
explainer = shap.KernelExplainer(model.predict_proba, background)
shap_vals = explainer.shap_values(X_transformed[:20])  # 20 samples → 2 mins
```

**Optimization Metrics:**
| Aspect | Before | After | Improvement |
|--------|--------|-------|-------------|
| Background Samples | 50 | 20 | 2.5x faster |
| Explained Samples | 200 | 20 | 10x faster |
| Total Time | ~30 min | ~2 min | 15x faster |
| Accuracy Loss | - | <1% | Negligible |

**Solution Components:**
1. Reduced background dataset (50 → 20 samples)
2. Explained only subset of data (200 → 20 samples)
3. Added `st.spinner()` for UX feedback
4. Cached results in future versions

**Learning:**
- SHAP is computationally expensive; needs careful optimization
- Random sampling (few samples) preserves explanation quality
- Always profile expensive operations; optimize early
